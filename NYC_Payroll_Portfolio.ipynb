# New York City Payroll


## Overview
Data from [NYC Open Data](https://data.cityofnewyork.us/City-Government/Citywide-Payroll-Data-Fiscal-Year-/k397-673e) representing New York City employee salary and overtime pay.  

#### Imports

```python
import pandas as pd
import numpy as np
```

#### Import data with dtype argument due to large file size

```python
nyc_full = pd.read_csv('NYC_Payroll_Data.csv', dtype={'Fiscal Year': 'int', 'Payroll Number': 'float', 'Agency Name': 'str',
                                                            'Last Name': 'str','First Name': 'str','Mid Init': 'str',
                                                            'Agency Start Date': 'str','Work Location Borough': 'str','Title Description': 'str',
                                                            'Leave Status as of June 30': 'str','Base Salary': 'float','Pay Basis': 'str',
                                                            'Regular Hours': 'float', 'Regular Gross Paid': 'float', 'OT Hours': 'float',
                                                            'Total OT Paid': 'float', 'Total Other Pay': 'float'})
```

#### Make backup copy of data before making changes

```python
nyc = nyc_full.copy()
```

#### Initial review of DataFrame

```python
nyc.info()
```

```python
nyc.head()
```

#### Transform characters to uppercase for consistency

```python
def upper_case(df):
    '''
    Takes as input the `nyc` DataFrame. 
    Then capitalizes each character in columns with "object" data type.
    '''
    # iterating over each column in the DataFrame
    for col in df.columns:
        # checking datatype of each column
        if df[col].dtype == 'object':
            # applying upper function on column
            df[col] = df[col].str.upper()

upper_case(nyc)
```

#### Create a DataFrame for 2021 fiscal year for further analysis

#### Create 'Total Gross Paid' column to total regular gross, total overtime and total other pay

```python
nyc_2021['Total Gross Paid'] = nyc_2021['Regular Gross Paid'] + nyc_2021['Total OT Paid'] + nyc_2021['Total Other Pay']
```

#### Calculate mean total gross pay

```python
nyc_2021['Total Gross Paid'].mean()
```

#### Identify number of different agencies represented in 2021

```python
nyc_2021['Agency Name'].nunique()
```

#### Determine which agencies had the highest / lowest median total gross pay

```python
# Create a SeriesGroupBy object for total gross pay by agency

agency_total_gross_pay = nyc_2021['Total Gross Paid'].groupby(nyc_2021['Agency Name'])
```

```python
agency_total_gross_pay
```

```python
# Calculate median total gross pay by agency and sort highest to lowest

agency_total_gross_pay.median().sort_values(ascending=False)
```

#### Calculate and compare highest individual total gross pay by agency

```python
agency_total_gross_pay.max().sort_values(ascending=False)
```

#### Identify the 5 highest paid employees (total gross pay) at Office of the Mayor

```python
# filter employees from Office of the Mayor

nyc_mayor_office = nyc_2021[nyc_2021["Agency Name"] == 'OFFICE OF THE MAYOR']
nyc_mayor_office.head()
```

```python
# Sort Office of the Mayor employees by highest total gross pay

nyc_mayor_office.sort_values(by = 'Total Gross Paid', ascending=False)
```

#### Calculate the Mayor's (Bill De Blasio) 2021 total gross pay 

```python
nyc_2021[
    (nyc_2021['First Name'] == 'BILL') 
    & (nyc_2021['Last Name'] == 'DE BLASIO')
    ]
```

#### Calculate average base salary per work location

```python
# Create a SeriesGroupBy object for average base salary by work location

avg_base_sal = nyc_2021['Base Salary'].groupby(nyc_2021['Work Location Borough'])
avg_base_sal
```

```python
# Calculate mean average base salary and sort lowest to highest

avg_base_sal.mean().sort_values(ascending = True)
```

#### Identify the highest base salary at the Washington, DC work location (the location with the highest mean base salary)

```python
# Filter employees at Washington, DC location

washdc = nyc_2021[nyc_2021["Work Location Borough"].isin(['WASHINGTON DC'])]


# Sort by base salary, highest to lowest

washdc.sort_values(by='Base Salary', ascending=False)
```
